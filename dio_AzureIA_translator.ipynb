```python

import requests
from docx import Document
import os

AZURE_TRANSLATOR_KEY = 'FjcUKw8r1WH3TSc2U4MuKYR9D5yc3uq1KjfORi652DnfIk6UTz9aJQQJ99BBACYeBjFXJ3w3AAAbACOGBQ1A'
AZURE_TRANSLATOR_LOCATION = 'eastus'
AZURE_TRANSLATOR_ENDPOINT = 'https://api.cognitive.microsofttranslator.com'

target_language = "pt-br"

def translate( text, target_language):
        url = f"{AZURE_TRANSLATOR_ENDPOINT}/translate?api-version=3.0&to={target_language}"
        headers = {
            'Ocp-Apim-Subscription-Key': AZURE_TRANSLATOR_KEY,
            'Ocp-Apim-Subscription-Region': AZURE_TRANSLATOR_LOCATION,
            'Content-type': 'application/json',
            'X-ClientTraceId': str(os.urandom(16))
        }
        body = [{
            'text': text
        }]
        response = requests.post(url, headers=headers, json=body)
        if response.status_code == 200:
            translated_text = response.json()[0]['translations'][0]['text']
            print(translated_text)
            return translated_text
        else:
            error_message = f"Erro ao traduzir: {response.status_code} - {response.text}"
            print(error_message)
            return error_message
```

```python
def translate_document(input_file_path):

    if not os.path.exists(input_file_path):
        print(f"Arquivo não encontrado: {input_file_path}")
        return

    document=Document(input_file_path)
    translated_paragraphs = []
    translated_doc = Document()

    for paragraphs in document.paragraphs:
        translator = translate(paragraphs.text, target_language)
        translated_paragraphs.append(translator)

    for line in translated_paragraphs:
      # print(line)
      translated_doc.add_paragraph(line)

    output_file_path = os.path.splitext(input_file_path)[0] + "_translated.docx"
    translated_doc.save(output_file_path)

    print(f"Tradução concluída! O novo arquivo foi salvo em: {output_file_path}")
```

```python
import requests
from bs4 import BeautifulSoup

def extract_text_from_url(url):
  response = requests.get(url)


  if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")

    for script_or_style in soup(["script", "style"]):
      script_or_style.decompose()

    text = soup.get_text(separator='  ')
    lines = (line.strip() for line in text.splitlines())
    chunks = (phrase.strip() for line in lines for phrase in line.split("  "))
    clean_text = '\n'.join(chunk for chunk in chunks if chunk)
    return clean_text
  else:
    error_message = f"Erro ao extrair: {response.status_code} - {response.text}"
    print(error_message)
    return error_message

```
